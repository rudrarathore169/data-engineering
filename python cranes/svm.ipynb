```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('pulsar_stars.csv')
df
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns
```

```python
df.columns=['IP Mean','IP Sd','IP Kurtosis','IP Skewness','DM-SNR Mean','DM-SNR Sd','DM-SNR Kurtosis',
           'DM-SNR Skewness','target_class']
```

```python
df.columns
```

```python
df['target_class'].value_counts()
```

```python
df.info()
```

```python
X=df.drop(['target_class'],axis=1)
y=df['target_class']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
X_train.shape,X_test.shape
```

```python
cols=X_train.columns
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train = pd.DataFrame(X_train,columns=[cols])
```

```python
X_test = pd.DataFrame(X_test,columns=[cols])
```

```python
X_train.describe()
```

```python
# import SVC Classifier
from sklearn.svm import SVC 
from sklearn.metrics import accuracy_score
svc = SVC()

svc.fit(X_train, y_train)
y_pred = svc.predict(X_test)
accuracy_score(y_test, y_pred)
print('Model accuracy scre with default hyperparameters: {0:0.4f}'.format(accuracy_score(y_test,y_pred)))
```

```python
svc = SVC(C=100.0)
svc.fit(X_train,y_train)
y_pred = svc.predict(X_test)

print('Model accuracy score with rbf kernel and c=100.0:{0:0.4f}'.
    format(accuracy_score(y_test,y_pred)))
```

```python
linear_svc = SVC(kernel='linear',C=1.0)
linear_svc.fit(X_train,y_train)
y_pred_test = linear_svc.predict(X_test)

print('Model accuracy score with linear kernel and c=1.0:{0:0.4f}'.
    format(accuracy_score(y_test,y_pred_test)))
```

```python
poly_svc =SVC(kernel='poly',C=1.0)
poly_svc.fit(X_train,y_train)
y_pred = poly_svc.predict(X_test)
print('Model accuracy score with polynomial kernel and c=1.0:{0:0.4f}'.
    format(accuracy_score(y_test,y_pred)))
```
