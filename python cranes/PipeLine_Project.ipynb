```python
#function to load the data(input file name), returns dataframe

def LoadDataset(file):
    d=pd.read_csv(file)
    return d

df_new=LoadDataset('titanicv1 - titanicv1.csv')
print(df_new)
```

```python
#function to understand dataframe(input dataframe) output is info, stats, datatype, rows & columns
def TheDataframe(d1):
    d1.info()
    d1.describe()
    d1.shape
    
TheDataframe(df_new)
```

```python
#function to check missing values(input df), return no of missing values in each column,
def CheckMissing(d1):
    a=d1.isnull().sum()
    return a
CheckMissing(df_new)
```
