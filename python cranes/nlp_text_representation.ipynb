```python
d1 = "the cat sat on the mat"
d2 = "the dog sat on the log"
d3 = "the cat chased the dog"
```

```python
docs = [d1,d2,d3]
```

```python
vocab = [w for d in docs for w in d.split()]
vocab=set(vocab)
vocab = sorted(vocab)
print(vocab)
```

```python
matrix=[]
for d in docs:
    row=[]
    for w in vocab:
        row.append(d.count(w))
    matrix.append(row)
print(matrix)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
x = cv.fit_transform(docs)
print(x.toarray())
print(cv.get_feature_names_out())
```

```python
# tfidf  implementation
d1 = "the cat sat on the mat"
d2 = "the dog sat on the log"
d3 = "the cat chased the dog"
docs = [d1,d2,d3]
```

```python
N = len(docs)
tokens = [doc.split() for doc in docs]
print(tokens)
```

```python
#example term t
term  = 'mat'
```

```python
#calculate the tf for above word
for i,toks in enumerate(tokens):
    tf = toks.count(term)/len(toks)
    print('tf:',tf)
    
    
```

```python
#compute the df and idf for the given term
import math
dfreq = sum(1 for toks in tokens if term in toks)
idf = math.log(N/dfreq)+1
print('idf:',idf)
```

```python
#compute tf idf
for i,toks in enumerate(tokens):
    tf = toks.count(term)/len(toks)
    tfidf = tf*idf
    print(tfidf)
```

```python
d1 = "the cat sat on the mat"
d2 = "the dog sat on the log"
d3 = "the cat chased the dog"
```

```python
docs = [d1,d2,d3]
```

```python
vocab = [w for d in docs for w in d.split()]
vocab=set(vocab)
vocab = sorted(vocab)
print(vocab)
```

```python
N = len(docs)
tokens = [doc.split() for doc in docs]
print(tokens)
```

```python
import pandas as pd
df = pd.DataFrame(index=range(len(tokens)), columns=vocab)
df1= pd.DataFrame(index=[0], columns=vocab)
print(df)
    
```

```python
for j in vocab:
    for i,toks in enumerate(tokens):
        tf = toks.count(j)/len(toks)
        df.loc[i, j] = tf
```

```python
print(df)
```

```python
import math
for j in vocab:
    dfreq = sum(1 for toks in tokens if j in toks)
    idf = math.log(N/dfreq)+1
    df1.loc[0,j]=idf
print(df1)
```

```python
tfidf = pd.DataFrame(index=range(len(tokens)), columns=vocab)
for j in vocab:
    for i,toks in enumerate(tokens):
        tfidf.loc[i,j] = df.loc[i,j]*df1.loc[0,j]

print(tfidf)
    
```
