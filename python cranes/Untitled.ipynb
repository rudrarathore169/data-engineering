```python
import re
import regex
```

```python
text = '''
        abcdefghijklmnopqrstuvwxyz
        ABCDEFGHIJKLMNOPWRSTUVWXYZ
        123456789
        abcd11234

        Hello CDOT Bangalore how are you?!
        . $ + | ()
        www.cdot.in
        7878496456
        196.132.100.23

        "Bangalore"
        \Cranes
        #333
        mallesh.vs@creanessoftware.com

        #978567

        'CDOT'
        '100.101a.300bc'

        Mr Damodharan
        Mr.Sivadas

        Mr Anoop Tiwari
        Mrs. Bhagaylakshmi

        Ms.Ramya
        ''' 
print(text)
```

```python
#start with creating object with the compile
P=re.compile(r'\.')#.is for all and \. for all dots 
M = P.finditer(text)
#find all print whatever is found
for m in M:
    print(m)
```

```python
#for all digits 
P = re.compile(r'\d')#\d is for all digits in the text
M = P.finditer(text)
#find all digits found
for m in M:
    print(m)
```

```python
# RE FUNCTION ALWAYS RETYRB NON OVERLAPING RESULTS
#find the specific digits btw 1 to 5

P=re.compile(r'[1-5]')
M = P.finditer(text)
#find all digits found
for m in M:
    print(m)
```

```python
# search for specific word
P = re.compile(r'Bangalore')# default case sensitive
M = P.finditer(text)
for m in M:
    print(m)
```

```python
# Case insensitive search
P = re.compile(r'bangalore',re.IGNORECASE)# case insensitive
M = P.finditer(text)
for m in M:
    print(m)
```

```python
# all lowercase
P = re.compile(r'[a-z]',)
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#search anything that matches in [lowercase] plus lowercase
P = re.compile(r'[a-z][a-z]')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#search anything that matches in [Uppercase] plus lowercase
P = re.compile(r'[A-Z][a-z]')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#search anything in [upper or lowercase]+lower alpha
P = re.compile(r'[A-Z+a-z][a-z]')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#[upper/lower]+[lower/upper/digits]
P = re.compile(r'[A-Z+a-z][a-z+A-Z+0-9]')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
# #+digits 0-5 plus 2 more digits
P = re.compile(r'#[0-5][0-9][0-9]')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
# #+digits 0-5 plus 2 more digits
P = re.compile(r'#[0-9][0-9][0-9]\s')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#find any six digit text stating with #
P = re.compile(r'#\d{6}\s')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
# find Hello
P =re.compile(r'[H][a-z]{4}\s')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#extract name with exactly 7 characters
P = re.compile(r'\b[a-z+A-Z]{7}(\s|\.)')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#extract all names starting with mr
p = re.compile(r'Mr(\.|\s)[A-Z][a-z]+')
M = p.finditer(text)
for m in M:
    print(m)
```

```python
#extract all names starting with mr and have lastname
P = re.compile(r'Mr(\.|\s)[A-Z][a-z]+\s[A-Z][a-z]+')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#make last name optional
P = re.compile(r'Mr(\.|\s)[A-Z][a-z]+(\s[A-Z][a-z]+)?')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#extract the ip addresses
P = re.compile(r'\d{3}\.\d{3}\.\d{3}\.\d{2}')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#extract Websites
P = re.compile(r'www\.\w+\.\w+')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
#exreact email
P = re.compile(r'[a-z+A-Z+0-9+.]+\@\w+\.\w+')
M = P.finditer(text)
for m in M:
    print(m)
```

```python
P = re.compile(r"'\D{4}'")
M = P.finditer(text)
for m in M:
    print(m)
```

```python
P = re.compile(r'"\D+"')
M = P.finditer(text)
for m in M:
    print(m)
```
