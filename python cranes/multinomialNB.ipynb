```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as snb
```

```python
data = pd.read_csv('spam.csv',encoding='latin')
data
```

```python
data['spam']=data['Category'].apply(lambda x:1 if x=='spam' else 0)
data
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(data.Message, data.spam,test_size=0.25,random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
b = CountVectorizer()
```

```python
x_train_count =b.fit_transform(x_train.values)
x_train_count.toarray()[:3]
```

```python
from sklearn.naive_bayes import MultinomialNB
mod1 = MultinomialNB()
mod1.fit(x_train_count,y_train)
```

```python
emails=["we have a annual meeting at 2.00pm",
       "you have 90% sales flat off on www.perfumes.com","offer!!!"]
```

```python
e_emails = b.transform(emails)
```

```python
e_emails
```

```python
x_test_count = b.transform(x_test)
```

```python
mod1.score(x_test_count,y_test)
```

```python
mod1.predict(e_emails)
```

```python
pred = mod1.predict(x_test_count)
```

```python
from sklearn.metrics import confusion_matrix,classification_report

print(confusion_matrix(y_test,pred))
print(classification_report(y_test,pred))
```
