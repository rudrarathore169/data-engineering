```
#build simple neural network for below data
import tensorflow as tf
import numpy as np
x = np.array([1,2,3,4,5])
y = np.array([5,6,7,8,9])
print(x,y)
```

```
from keras import metrics
#build the NN architechture with keras
from tensorflow import keras
# to define model type sequential neurons connected one after other
from keras.models import Sequential
#to add input and output neurons to model
from keras.layers import Dense
#define the architecture
model = Sequential()
model.add(Dense(units=1,input_shape=(1,)))#units =1 means output neuron is 1
model.summary()
#define optimizer and matrix
model.compile(optimizer='sgd',loss=['mean_squared_error'])
```

```
model.fit(x,y,epochs=1000)
```

```
#PRINT THE MODEL WEIGHTS
print(model.get_weights())
```

```
# predict on sample data
print(model.predict(np.array([3])))
```

```
# save the model
model.save('https://github.com/rudrarathore169/data-engineering/tree/main/Projects/mymodel.h5')
```
